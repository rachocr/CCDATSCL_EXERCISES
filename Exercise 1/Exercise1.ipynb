## C. Cleaning



Check for missing values

```python
df.isnull().sum()
```

Check for outliers in `minimum_nights`

```python
df['minimum_nights'].describe()
```

Check for invalid values (e.g., price ≤ 0)

```python
df[df['price'] <= 0].shape[0]
```

Apply a missing-value strategy (drop, fill, or flag)


```python
df['reviews_per_month'] = df['reviews_per_month'].fillna(0.0)

if 'Unknown' not in df['host_name'].cat.categories:
    df['host_name'] = df['host_name'].cat.add_categories('Unknown')

df['host_name'] = df['host_name'].fillna('Unknown')
df['name'] = df['name'].fillna('Unknown Listing Name')

print("Missing value imputation complete. Check the new null counts.")
print(df.isnull().sum())
```

Apply an outlier-handling strategy

```python
initial_rows = df.shape[0]
df = df[df['price'] > 0]
rows_removed_price = initial_rows - df.shape[0]
print(f"Removed {rows_removed_price} rows with non-positive prices (price <= 0).")

initial_rows_after_price_check = df.shape[0]
df = df[df['minimum_nights'] <= 365]
rows_removed_nights = initial_rows_after_price_check - df.shape[0]
print(f"Removed {rows_removed_nights} rows where minimum_nights exceeded 365.")

print(f"\nFinal DataFrame shape after outlier and invalid value cleaning: {df.shape}")
```

## D. Enriching

Create at least 3 new variables:
- price_per_minimum_night
- review_rate_per_month (combine ratings + frequency)
- is_expensive (boolean flag based on price threshold)


```python
df['price_per_minimum_night'] = df['price'] / df['minimum_nights']

df['review_rate_per_month'] = df['number_of_reviews'] * df['reviews_per_month']

price_75th_percentile = df['price'].quantile(0.75)
df['is_expensive'] = (df['price'] > price_75th_percentile)

print(f"Price threshold for 'is_expensive' (75th Percentile): ${price_75th_percentile:.2f}")
print("New features created. Displaying head with new columns:")
df[['price', 'minimum_nights', 'price_per_minimum_night', 'number_of_reviews', 'reviews_per_month', 'review_rate_per_month', 'is_expensive']].head()
```

Create 1 aggregated summary table

(e.g. average price per neighborhood number of hosts per borough)

```python
neighbourhood_summary = df.groupby('neighbourhood_group').agg(
    average_price=('price', 'mean'),
    median_price=('price', 'median'),
    total_listings=('id', 'count'),
    total_unique_hosts=('host_id', 'nunique')
).reset_index()

print("Aggregated Summary Table (Price and Host Count per Borough):")
neighbourhood_summary
```

## E. Validating

Create at least 3 validation checks
- Price must be positive.
- Host ID should not be null.
- Minimum nights must be ≥ 1.
- Generate a validation report:
- Count of rows violating each rule.

Decide if invalid rows (if any) should be removed or corrected?

```python
# Create validation checks
price_violation = df['price'] <= 0
host_id_violation = df['host_id'].isnull()
min_nights_violation = df['minimum_nights'] < 1

# Generate the Validation Report
validation_report = pd.DataFrame({
    'Rule Violated': [
        'Price must be positive',
        'Host ID should not be null',
        'Minimum nights must be ≥ 1'
    ],
    'Violating Rows Count': [
        price_violation.sum(),
        host_id_violation.sum(),
        min_nights_violation.sum()
    ]
})

print("--- VALIDATION REPORT ---")
print(validation_report)
```

## F. Publishing

Select final columns for publication.



```python
final_columns = [
    'id', 'name', 'host_id', 'host_name', 'neighbourhood_group', 'neighbourhood',
    'latitude', 'longitude', 'room_type', 'price', 'minimum_nights',
    'number_of_reviews', 'last_review', 'reviews_per_month',
    'calculated_host_listings_count', 'availability_365',
    'price_per_minimum_night', 'review_rate_per_month', 'is_expensive'
]

df_final = df[final_columns].copy()

print("Final dataset prepared with 19 columns, including 3 new enriched features.")
print(df_final.head(3))
```

Export cleaned dataset as "cleaned_airbnb.csv"

Export summary reports (e.g., validation or aggregations) as CSV.
