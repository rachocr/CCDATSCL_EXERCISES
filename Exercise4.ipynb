```python
import kagglehub
import os
import pandas as pd

import plotly.express as px
import seaborn as sns
import matplotlib.pyplot as plt

# Download latest version
path = kagglehub.dataset_download("imdevskp/corona-virus-report")

print("Path to dataset files:", path)
```

```python
if os.path.isdir(path):
  print(True)

contents = os.listdir(path)
contents

mydataset = path + "/" + contents[0]
mydataset


df = pd.read_csv(mydataset)
```

## A. Time-Based Visualizations

1. Global Trend `(5 pts)`

Aggregate the data by Date and create a line chart showing the global number of confirmed COVID-19 cases over time.

```python
df['Country/Region'].unique()
```

```python
df_confirmed = df.groupby('Date')['Confirmed'].sum().reset_index()

fig = px.line(df_confirmed, x='Date', y='Confirmed',
              title='Global Trend of Confirmed COVID-19 Cases',
              labels={'Confirmed': 'Confirmed Cases'})
fig.show()
```

2. Country-Level Trends `(5 pts)`

Select three countries and visualize their confirmed case counts over time on the same plot.

```python
selected_countries = ['China', 'Philippines', 'Singapore']
df_countries = df[df['Country/Region'].isin(selected_countries)]

df_countries_trend = df_countries.groupby(['Date', 'Country/Region'])['Confirmed'].sum().reset_index()

fig = px.line(df_countries_trend,
              x='Date',
              y='Confirmed',
              color='Country/Region',
              title='Confirmed Cases Trend: China vs Philippines vs Singapore')

fig.show()
```

3. Active vs Recovered `(5 pts)`

For a selected country, create a line chart showing Active and Recovered cases over time.

```python
selected_country = 'China'
df_country_cases = df[df['Country/Region'] == selected_country]

df_trend_active_recovered = df_country_cases.groupby('Date')[['Active', 'Recovered']].sum().reset_index()

fig = px.line(df_trend_active_recovered,
              x='Date',
              y=['Active', 'Recovered'],
              title=f'Active vs. Recovered COVID-19 Cases in {selected_country}',
              labels={'value': 'Number of Cases', 'variable': 'Case Type'})
fig.show()
```

```python
selected_country = 'Philippines'
df_country_cases = df[df['Country/Region'] == selected_country]

df_trend_active_recovered = df_country_cases.groupby('Date')[['Active', 'Recovered']].sum().reset_index()

fig = px.line(df_trend_active_recovered,
              x='Date',
              y=['Active', 'Recovered'],
              title=f'Active vs. Recovered COVID-19 Cases in {selected_country}',
              labels={'value': 'Number of Cases', 'variable': 'Case Type'})
fig.show()
```

```python
selected_country = 'Singapore'
df_country_cases = df[df['Country/Region'] == selected_country]

df_trend_active_recovered = df_country_cases.groupby('Date')[['Active', 'Recovered']].sum().reset_index()

fig = px.line(df_trend_active_recovered,
              x='Date',
              y=['Active', 'Recovered'],
              title=f'Active vs. Recovered COVID-19 Cases in {selected_country}',
              labels={'value': 'Number of Cases', 'variable': 'Case Type'})
fig.show()
```

## B: Comparative Visualizations

4. Country Comparison `(5 pts)`

Using data from a single date, create a bar chart showing the top 10 countries by confirmed cases.

```python
selected_date = df['Date'].max()
df_date = df[df['Date'] == selected_date]

df_date.groupby('Country/Region')['Confirmed'].sum().sort_values(ascending=False)
```

```python
top_countries = df_date.groupby('Country/Region')['Confirmed'].sum().nlargest(10).reset_index()

fig = px.bar(top_countries, x='Country/Region', color='Country/Region', y='Confirmed', title=f'Top 10 Countries by Confirmed Cases', labels={'Confirmed': 'Confirmed Cases', 'Country/Region': 'Country'})
fig.show()
```

5. WHO Region Comparison `(5 pts)`

Aggregate confirmed cases by WHO Region and visualize the result using a bar chart.

```python
df.groupby('WHO Region')['Confirmed'].sum().sort_values(ascending=False)
```

```python
df_who_region_cases = df.groupby('WHO Region')['Confirmed'].sum().reset_index()

fig = px.bar(df_who_region_cases,x='WHO Region', y='Confirmed', color='WHO Region', title='Confirmed Cases by WHO Region', labels={'Confirmed': 'Total Confirmed Cases', 'WHO Region': 'WHO Region'})
fig.show()
```

## C. Geographic Visualization

6. Geographic Spread `(10 pts)`

Using Latitude and Longitude, create a map-based visualization showing confirmed cases for a selected date.

```python
selected_date = df['Date'].max()
df_latest = df[df["Date"] == selected_date]

fig = px.choropleth(
    df_latest,
    locations="Country/Region",
    locationmode="country names",
    color="Confirmed",
    hover_name="Country/Region",
    projection="natural earth",
    title=f"Global Confirmed Cases in {selected_date}"
)

fig.show()
```

7. Regional Clustering `(15 pts)`

Create a visualization that shows how confirmed cases are distributed geographically within a single WHO Region.

```python
target_region = 'Americas'
df_region = df[(df["Date"] == selected_date) & (df["WHO Region"] == target_region)]

fig = px.choropleth(
    df_region,
    locations="Country/Region",
    locationmode="country names",
    color="Confirmed",
    hover_name="Country/Region",
    projection="natural earth",
    title=f"Confirmed Cases Distribution in {target_region} Region",
)

fig.show()
```
